```python
import random

def generar_matrices(N):
    matriz = []
    for i in range(N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1,9))
        matriz.append(fila)
    return matriz

N = 5

matriz_aleatoria = generar_matrices(N)

for fila in matriz_aleatoria:
    print(fila)

def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ''.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)

```

```python
print('matriz de 10x 10')
m_10 = generar_matrices(20)
imprimir_matriz(m_10)
print("*")
m_20 = generar_matrices(30)
imprimir_matriz(m_20)
```
